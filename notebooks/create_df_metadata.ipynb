```python
import pandas as pd
```

```python
df_cancer = pd.read_csv("../data/processed/tcga_brca_raw_counts_CANCER.csv", index_col=0)
df_control = pd.read_csv("../data/processed/tcga_brca_raw_counts_CONTROL.csv", index_col=0)
```

## Create Metada Dataframe

```python
df_md_cancer = df_cancer.copy()
df_md_control = df_control.copy()
```

```python
# Create column Condition
df_md_control["Condition"] = "Healthy"
df_md_cancer["Condition"] = "Cancer"

# Keep only the Condition column
df_md_control = df_md_control[["Condition"]]
df_md_cancer = df_md_cancer[["Condition"]]
```

```python
df_metadata = pd.concat([df_md_control, df_md_cancer],axis=0,ignore_index=False)
```

```python
df_metadata
```

```python
df_metadata.to_csv("../data/processed/metadata_merged.csv")
```

## Merge Raw Counts Dataframe

```python
df_raw_cancer = df_cancer.copy()
df_raw_control = df_control.copy()
```

```python
df_raw_merged = pd.concat([df_raw_control, df_raw_cancer],axis=0,ignore_index=False)
```

```python
df_raw_merged.to_csv("../data/processed/raw_merged.csv")
```
